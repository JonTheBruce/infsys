## 4. Load pretrained BERT

```python

# Load the BERT tokenizer and model.
# This may take a few minutes the first time you run it.

model_name = "bert-base-uncased"

tokenizer = BertTokenizer.from_pretrained(model_name)
bert_model = BertModel.from_pretrained(model_name)

# Put BERT in evaluation mode because we are only using it to create embeddings.
bert_model.eval()
```


## 5. Create BERT embeddings

BERT converts each alert description into a sequence of token embeddings.

For this assignment, we will use **mean pooling**: we average the token embeddings to create one vector for each alert description.

The output will be a matrix where:

- Each row represents one alert.
- Each column represents one dimension of the BERT embedding.


```python

def get_bert_embeddings(texts, tokenizer, model, max_length=96, batch_size=8):
    """
    Convert a list of text strings into BERT embeddings using mean pooling.
    """
    all_embeddings = []

    for i in range(0, len(texts), batch_size):
        batch_texts = texts[i:i + batch_size]

        encoded = tokenizer(
            batch_texts,
            padding=True,
            truncation=True,
            max_length=max_length,
            return_tensors="pt"
        )

        with torch.no_grad():
            outputs = model(**encoded)

        token_embeddings = outputs.last_hidden_state
        attention_mask = encoded["attention_mask"]

        # Expand the attention mask so it can be multiplied with token embeddings.
        input_mask_expanded = attention_mask.unsqueeze(-1).expand(token_embeddings.size()).float()

        # Sum token embeddings and divide by the number of non-padding tokens.
        sum_embeddings = torch.sum(token_embeddings * input_mask_expanded, dim=1)
        sum_mask = torch.clamp(input_mask_expanded.sum(dim=1), min=1e-9)
        mean_embeddings = sum_embeddings / sum_mask

        all_embeddings.append(mean_embeddings.cpu().numpy())

    return np.vstack(all_embeddings)


texts = df["alert_text"].tolist()

bert_embeddings = get_bert_embeddings(texts, tokenizer, bert_model)

print("BERT embedding matrix shape:", bert_embeddings.shape)
```

## 7. Prepare data for regression


We will compare three feature sets:

1. **Structured features only**
2. **BERT embeddings only**
3. **Structured features + BERT embeddings**

The target variable is `severity_score`.


```python

structured_features = [
    "failed_login_count",
    "unusual_geo",
    "privilege_escalation",
    "data_exfiltration",
    "malware_detected",
    "asset_criticality"
]

X_structured = df[structured_features].values
X_bert = bert_embeddings
y = df["severity_score"].values

# Scale structured features before combining them with BERT embeddings.
scaler = StandardScaler()
X_structured_scaled = scaler.fit_transform(X_structured)

X_combined = np.hstack([X_structured_scaled, X_bert])

print("Structured feature shape:", X_structured_scaled.shape)
print("BERT feature shape:", X_bert.shape)
print("Combined feature shape:", X_combined.shape)
```

```python

# Split all three feature sets using the same train/test split.

X_train_struct, X_test_struct, y_train, y_test = train_test_split(
    X_structured_scaled, y, test_size=0.25, random_state=42
)

X_train_bert, X_test_bert, _, _ = train_test_split(
    X_bert, y, test_size=0.25, random_state=42
)

X_train_combined, X_test_combined, _, _ = train_test_split(
    X_combined, y, test_size=0.25, random_state=42
)

print("Training observations:", len(y_train))
print("Testing observations:", len(y_test))
```

## 8. Define an evaluation function

```python

def evaluate_regression_model(model, X_train, X_test, y_train, y_test, model_name):
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    mae = mean_absolute_error(y_test, predictions)
    rmse = np.sqrt(mean_squared_error(y_test, predictions))
    r2 = r2_score(y_test, predictions)

    return {
        "model": model_name,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    }, predictions
```

## 9. Train and evaluate baseline and regression models

```python

results = []

# Baseline model: always predicts the average severity score in the training data.
baseline = DummyRegressor(strategy="mean")
baseline_result, baseline_predictions = evaluate_regression_model(
    baseline,
    X_train_struct,
    X_test_struct,
    y_train,
    y_test,
    "Baseline Mean Predictor"
)
results.append(baseline_result)

# Ridge Regression with structured features only.
ridge_struct = Ridge(alpha=1.0)
ridge_struct_result, ridge_struct_predictions = evaluate_regression_model(
    ridge_struct,
    X_train_struct,
    X_test_struct,
    y_train,
    y_test,
    "Ridge Regression: Structured Features Only"
)
results.append(ridge_struct_result)

# Ridge Regression with BERT embeddings only.
ridge_bert = Ridge(alpha=10.0)
ridge_bert_result, ridge_bert_predictions = evaluate_regression_model(
    ridge_bert,
    X_train_bert,
    X_test_bert,
    y_train,
    y_test,
    "Ridge Regression: BERT Embeddings Only"
)
results.append(ridge_bert_result)

# Ridge Regression with structured features + BERT embeddings.
ridge_combined = Ridge(alpha=10.0)
ridge_combined_result, ridge_combined_predictions = evaluate_regression_model(
    ridge_combined,
    X_train_combined,
    X_test_combined,
    y_train,
    y_test,
    "Ridge Regression: Structured + BERT"
)
results.append(ridge_combined_result)

results_df = pd.DataFrame(results)
results_df
```

Why should cybersecurity analysts not rely only on an AI-generated severity score? AI models are trained on historical correlations and can be misled by novel attack vectors, incomplete data, or out-of-distribution patterns (such as zero-day exploits). Over-relying on an automated score can lead to a dangerous false sense of security where critical, unique indicators of a sophisticated breach are missed simply because the model flagged them as low-priority.

What types of alerts might be unfairly underestimated by a model trained on historical data? Alerts containing novel techniques, subtle low-and-slow tactics, or zero-day vulnerabilities that have no representative examples in the training set will likely be underestimated. Additionally, highly targeted attacks (like spear-phishing tailored to a specific executive) may appear linguistically similar to normal, low-severity communications, causing the model to misclassify them as low-risk.

What information should be shown to a human analyst so they can understand why an alert received a high or low score? To make the model's prediction transparent, the analyst should be shown:

Feature Contributions / Importances: Which structured features (e.g., high asset criticality, privilege escalation detected) contributed most to the score.
Linguistic Highlights: The specific words or semantic concepts from the unstructured text (using techniques like Integrated Gradients or LIME) that triggered the model.
Historical Context / Similar Cases: A list of past, similar alerts that were handled by human analysts to serve as reference points.
How could an organization monitor this model after deployment? Organizations can monitor the model by:

Tracking Performance Metrics: Continually comparing predicted risk levels/severity scores against the final ground-truth severity determinations made manually by Tier 2/3 analysts (looking for drift in Precision/Recall or MAE).
Data Drift Detection: Monitoring the incoming text and feature distributions to ensure the model isn't being exposed to entirely new environments, jargon, or attack categories.
Feedback Loops: Allowing analysts to explicitly flag and correct misclassified alerts, which are then used to update the training dataset.
What are the advantages and disadvantages of using BERT embeddings instead of simpler methods such as TF-IDF?

Advantages: BERT captures deep contextual relationships and polysemy (the same word meaning different things in different contexts) and can leverage synonymy (mapping different words with the same meaning to a similar vector space). Unlike TF-IDF, it captures the underlying semantics of security descriptions rather than just counting exact word matches.
Disadvantages: BERT is computationally expensive, requires massive GPU resources, and maps text to a very high-dimensional space (768 dimensions), which easily leads to overfitting and noise on small datasets. TF-IDF is much faster to compute, completely transparent to interpret, and often performs better when data is highly limited.
